#HYPOTHESIS TESTING

STEP 3 : HYPOTHESIS TESTING

3.1 Define hypothesis

Business Hypothesis



H0 (Null Hypothesis):
There is no significant difference in release years between TV-MA and PG content on Netflix.

H1 (Alternative Hypothesis):
There is a significant difference in release years between TV-MA and PG content.

```python
import pandas as pd
from scipy.stats import ttest_ind
```

```python
df = pd.read_csv("netflix_cleaned.csv")
df.head()
```

3.2 Perform test

```python
df = df.dropna(subset=['rating', 'release_year'])
```

```python
tv_ma = df[df['rating'] == 'TV-MA']['release_year']
pg = df[df['rating'] == 'PG']['release_year']
```

```python
print("TV-MA count:", len(tv_ma))
print("PG count:", len(pg))
```

```python
t_stat, p_value = ttest_ind(tv_ma, pg, equal_var=False)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
print("Mean TV-MA:", tv_ma.mean())
print("Mean PG:", pg.mean())
```

Case 1: If p-value > 0.05

Conclusion:

The p-value is greater than the significance level (0.05), so we fail to reject the null hypothesis.
This indicates that there is no statistically significant difference in the release years of TV-MA and PG content on Netflix.

Case 2: If p-value < 0.05

Conclusion:

The p-value is less than the significance level (0.05), so we reject the null hypothesis.
This indicates that there is a statistically significant difference in the release years of TV-MA and PG content on Netflix.
