#🔹 Step 2: Data Cleaning

2.1 Check Missing Values

```python
df.isnull().sum()
```

2.2 Handle Missing Values

```python
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Not Available')
df['country'] = df['country'].fillna('Unknown')
df['rating'] = df['rating'].fillna('Not Rated')
```

2.3 Remove Duplicates

```python
df.duplicated().sum()
```

2.4 Convert Date Column

```python
df['date_added'] = df['date_added'].str.strip()
```

```python
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
```

```python
df['date_added'].isnull().sum()
```

```python
df['date_added'] = df['date_added'].ffill()
```

#Step 4: Exploratory Data Analysis (EDA)

```python
import matplotlib.pyplot as plt
```

4.1 Distribution of Content Type (Movies vs TV Shows)

```python
counts = df['type'].value_counts()

plt.figure()
counts.plot(kind='barh')

plt.title("Distribution of Content Type (Movies vs TV Shows)")
plt.xlabel("Count")
plt.ylabel("Content Type")

# Add value labels
for index, value in enumerate(counts):
    plt.text(value, index, str(value))

plt.show()
```

```python
country_counts = df[df['country'] != 'Unknown']['country'].value_counts().head(10)
```

4.2 Top 10 Countries Producing Netflix Content

```python
plt.figure()

country_counts.plot(kind='bar')

plt.title("Top 10 Countries Producing Netflix Content")
plt.xlabel("Country")
plt.ylabel("Count")

# Rotate labels for better readability
plt.xticks(rotation=45)

# Add values on top
for i, v in enumerate(country_counts):
    plt.text(i, v, str(v))

plt.show()
```

```python
rating_counts = df[df['rating'] != 'Not Rated']['rating'].value_counts().head(10)
```

4.3 Top 10 Most Common Ratings on Netflix

```python


plt.figure()

rating_counts.plot(kind='bar')

plt.title("Top 10 Most Common Ratings on Netflix")
plt.xlabel("Rating")
plt.ylabel("Count")

plt.xticks(rotation=45)

# Add values on bars
for i, v in enumerate(rating_counts):
    plt.text(i, v, str(v))

plt.show()
```

```python
df['year_added'] = df['date_added'].dt.year
```

```python
year_counts = df['year_added'].value_counts().sort_index()
```

4.4 Content Added to Netflix Over Time

```python
plt.figure()

year_counts.plot(kind='line', marker='o')

plt.title("Content Added to Netflix Over Time")
plt.xlabel("Year")
plt.ylabel("Number of Titles Added")

# Add values on points
for x, y in zip(year_counts.index, year_counts.values):
    plt.text(x, y, str(y))

plt.show()
```
